```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt

PROC = "../data/processed/"
TABL = "../outputs/tables/"
os.makedirs(TABL, exist_ok=True)

def rotate_upright(img):
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    inv = cv2.bitwise_not(gray)
    thr = cv2.threshold(inv, 0, 255, cv2.THRESH_BINARY+cv2.THRESH_OTSU)[1]
    coords = np.column_stack(np.where(thr > 0))
    angle = 0.0
    if coords.size > 0:
        angle = cv2.minAreaRect(coords)[-1]
        angle = -(90 + angle) if angle < -45 else -angle
    (h,w) = img.shape[:2]
    M = cv2.getRotationMatrix2D((w//2, h//2), angle, 1.0)
    return cv2.warpAffine(img, M, (w, h), flags=cv2.INTER_CUBIC, borderMode=cv2.BORDER_REPLICATE)

def detect_and_crop_tables(image_path, out_dir, min_w=200, min_h=120):
    orig = cv2.imread(image_path)
    img = rotate_upright(orig)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    inv = cv2.bitwise_not(gray)
    thr = cv2.threshold(inv, 0, 255, cv2.THRESH_BINARY+cv2.THRESH_OTSU)[1]

    # detect horizontal + vertical lines
    h_kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (40,1))
    v_kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (1,40))
    h_lines = cv2.morphologyEx(thr, cv2.MORPH_OPEN, h_kernel, iterations=2)
    v_lines = cv2.morphologyEx(thr, cv2.MORPH_OPEN, v_kernel, iterations=2)
    mask = cv2.add(h_lines, v_lines)

    cnts,_ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    boxes = []
    for c in cnts:
        x,y,w,h = cv2.boundingRect(c)
        if w>=min_w and h>=min_h:
            boxes.append((x,y,w,h))

    boxes = sorted(boxes, key=lambda b:(b[1], b[0]))  # top-to-bottom

    page_base = os.path.splitext(os.path.basename(image_path))[0]
    for i,(x,y,w,h) in enumerate(boxes, start=1):
        cv2.rectangle(img, (x,y), (x+w,y+h), (0,255,0), 2)
        crop = img[y:y+h, x:x+w]
        outp = os.path.join(out_dir, f"{page_base}_table_{i}.png")
        cv2.imwrite(outp, crop)

    # show preview
    rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    plt.figure(figsize=(12,10)); plt.imshow(rgb); plt.axis('off'); plt.show()
    return boxes
```

```python
total_crops = 0
for fn in os.listdir(PROC):
    if fn.lower().endswith((".png",".jpg",".jpeg")):
        fp = os.path.join(PROC, fn)
        boxes = detect_and_crop_tables(fp, TABL)
        print(f"[TABLE] {fn}: {len(boxes)} region(s)")
        total_crops += len(boxes)
print(f"✅ Table detection done. Crops saved in {TABL} (total: {total_crops})")
```
